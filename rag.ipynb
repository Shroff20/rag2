```python


import pandas as pd


def _get_simplified_document_df():

    results = DS.collection.get(where = {'chunk_idx':-1})

    documents = results["documents"]
    truncated_documents = [s[:100000000000000] + '... (truncated)' for s in documents]

    df_meta = pd.DataFrame.from_dict(results["metadatas"])
    df_ids = pd.DataFrame({"id": results["ids"]})
    df_documents = pd.DataFrame({"document": truncated_documents})
    df_results = pd.concat(
        [df_ids, df_meta, df_documents],
        axis=1,
    )

    df_results = df_results.drop(columns = ['chunk_idx', 'hash', 'source_type', 'N_chunks', 'source_id', 'pca'])

    df_results = shroff_rag._reorder_cols(df_results, ['basename', 'document'])

    return df_results
df_results = _get_simplified_document_df()



print(df_results)

```
