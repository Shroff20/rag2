```python
device = "cuda"
database_folder = "./database_data"


DS = database.DataStore(device, database_folder)
DS.clear_collections()

DS.create_collection()

print(DS)
```

```python
glob_pattern = os.path.join('.', 'data', '*.pdf')
files = glob.glob(glob_pattern)

for file in files:
   id, document, metadata = parsers.parse_pdf(file)
   DS.add([id,], [document,], [metadata,])
   

```
