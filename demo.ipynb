```python
import pandas as pd
import ast




def results_to_df(results, document_length_limit = None):

    d = {}
    if 'distances' in results.keys():
        d['distance'] = results['distances']
    if 'ids' in results.keys():
        d['id'] = results['ids']
    if 'documents' in results.keys():  
        d['document'] = results['documents']
    if 'embeddings' in results.keys():
        d['embedding'] = [row for row in results['embeddings']]
    df = pd.DataFrame.from_dict(d)

    if 'metadatas' in results.keys():
        df_metadata = pd.DataFrame(results['metadatas'], index = range(len(df))).drop(columns = 'id')
        df = pd.concat([df, df_metadata], axis = 1).reset_index(drop = True)

    datetime_format =  r"%Y-%m-%d %H:%M:%S.%f"
    type_converter = {}
    type_converter['creation_date'] = lambda x:  pd.to_datetime(x, format = datetime_format)
    type_converter['upload_date'] = lambda x:  pd.to_datetime(x, format = datetime_format)
    type_converter['modification_date'] = lambda x:  pd.to_datetime(x, format = datetime_format)
    type_converter['page_lengths'] = lambda x:  x.apply(ast.literal_eval)
    type_converter['document'] = lambda x:  x.str[:document_length_limit]

    for col in df.columns:
        if col in type_converter.keys():
            df[col] = type_converter[col](df[col])

    return df


results = VD.collection.get(
        limit=10,
        offset=0,
        include=['documents', 'metadatas', 'embeddings'] # Specify what to include
    )


df = results_to_df(results, document_length_limit= 6)

df
```
